```python
results=pd.read_csv('df_n4.csv')
```

```python
import ast

def to_list(x):
    if isinstance(x, str):
        return ast.literal_eval(x)
    return x

results["num_comments"] = (
    results["comment_ids"]
    .apply(to_list)
    .apply(len)
)
```

```python
results[5:].head(20)
```

```python
import pandas as pd
import ast

df = results.copy()

# Convert string representations of lists to Python lists
list_cols = [
    "comment_ids",
    "comments",
    "generation_systems",
    "categories",
    "severities",
]

for col in list_cols:
    df[col] = df[col].apply(
        lambda x: ast.literal_eval(x) if isinstance(x, str) else x
    )

# One row = one generated comment
exploded = df.explode(
    ["comment_ids",
     "comments",
     "generation_systems",
     "categories",
     "severities"],
    ignore_index=True
)
```

```python
results.head(20)
```

```python
severity_dist = (
    exploded
    .groupby(["generation_systems","severities"])
    .size()
    .unstack(fill_value=0)
)
severity_dist
```

```python
# categories agreement 
agreement = (
    exploded
    .groupby("cluster_id")["categories"]
    .nunique()
)

agreement_rate = (agreement==1).mean()

print(f"Agreement = {agreement_rate:.2%}")
```

```python
# serevrity agreement 
agreement = (
    exploded
    .groupby("cluster_id")["severities"]
    .nunique()
)

agreement_rate = (agreement==1).mean()

print(f"Agreement = {agreement_rate:.2%}")
```

```python
cluster_sizes = (
    exploded
    .groupby("cluster_id")
    .size()
)

print(cluster_sizes.describe())
```

```python
# Which categories produce the biggest consensus

tmp = exploded.copy()

tmp["cluster_size"] = (
    tmp.cluster_id.map(cluster_sizes)
)

tmp.groupby("categories")["cluster_size"].mean().sort_values(ascending=False)
```

```python
from itertools import combinations
from collections import defaultdict

# Which systems agree the most
overlap = defaultdict(int)

clusters = exploded.groupby("cluster_id")

for _,g in clusters:

    systems = sorted(g.generation_systems.unique())

    for s1,s2 in combinations(systems,2):

        overlap[(s1,s2)] +=1

overlap
```

```python
# Clusters found only by one system
unique_clusters = (
    exploded
    .groupby("cluster_id")
    .generation_systems
    .nunique()
)

single = unique_clusters[unique_clusters==1].index

exploded[
    exploded.cluster_id.isin(single)
].generation_systems.value_counts()
```
