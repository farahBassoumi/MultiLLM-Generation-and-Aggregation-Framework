```python
import pandas as pd
df=pd.read_csv('df_n5.csv')
df6=pd.read_csv('df_n6_min.csv')
```

```python
df = df.merge(
    df6[["patch_id", "matching_cluster_ids"]],
    on="patch_id",
    how="left"
)

df["match_human_comment"] = df.apply(
    lambda row: row["cluster_id"] in row["matching_cluster_ids"],
    axis=1
)
```

```python
df["is_useful"] = df["relevance_score"] >= 3
```

```python
df.info()
```

```python
df.groupby("relevance_score")["num_comments"].agg(
    count="count",
    avg_num_comments="mean"
)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    cohen_kappa_score
)

# Ground truth
y_true = df["match_human_comment"]

# Predictions
y_pred = df["is_useful"]

# Metrics
accuracy = accuracy_score(y_true, y_pred)
precision = precision_score(y_true, y_pred, zero_division=0)
recall = recall_score(y_true, y_pred, zero_division=0)
f1 = f1_score(y_true, y_pred, zero_division=0)

print(f"Accuracy:  {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall:    {recall:.4f}")
print(f"F1-score:  {f1:.4f}")
```
